# Understanding Document Structure

```python
from langchain_core.documents import Document
```

```python
doc = Document(
    page_content = "This is a RAG sample document, whcih is supposed to be a text",
    metadata = {
        "author":"Ankit",
        "date_created":"2026-04-27",
        "file_type": "text"
        # metadata will help in filtering for semantic search - "What is the best quote by Author Ankit" - here filter would be done using metada
    }
)
print(doc)
type(doc)
#calling the object
#doc.metadata
#doc.page_content


#This is how we will turn any kind of source (pdf, doc, json, GoogleDrive, CSV, SQL ...) into a Document Structure - while loading the initial document
```

# Data Ingestion - Loading and Parsing Documents

```python
# Text Loader
```

```python
# Directory loader - Load all files from a directory
from langchain_community.document_loaders import DirectoryLoader, TextLoader

dir_load = DirectoryLoader(
    path="text_directory/",
    glob="**/*.txt", #mathcing the file pattern
    loader_cls= TextLoader, #can also provide a list if multiple file types are expected [TextLoade, PDFLoader, CSVLoader....]
    show_progress=True,
)
dir_load.load()

#Call object
# dir_load[0]
```

```python
# PDF Loader
from langchain_community.document_loaders import PyMuPDFLoader ,PyPDFLoader

pdf_loader = DirectoryLoader(
    path = "pdf_directory",
    glob="**/*.pdf",
    loader_cls=PyMuPDFLoader
)
pdf_loader.load()
```
